```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df=pd.read_csv('dm_office_sales.csv')
df=pd.DataFrame(df)
df
fig=plt.figure(figsize=(10,6),dpi=100)
sns.countplot(data=df,x='Division',hue='Division').set(title='Count of Each Department',xlabel='Department',ylabel='Count')
df['Division'].value_counts()
```

```python
df
fig=plt.figure(figsize=(10,6),dpi=100)
sns.countplot(data=df,x='Level Education',hue='Division',palette='Set2').set(title='Count of Each Department',xlabel='Department',ylabel='Count')
```

```python
sns.barplot(data=df,x='Level Education',y='Salary',estimator=np.mean,errorbar='sd',palette='Set2',hue='Division').set(title='Average Salary by Education Level',xlabel='Education Level',ylabel='Average Salary')
plt.legend(bbox_to_anchor=(1.05,1),loc=2,borderaxespad=0.)
```
