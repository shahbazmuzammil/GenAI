```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('dm_office_sales.csv')
df=pd.DataFrame(df)
fig=df.describe().transpose()
sns.histplot(data=fig,x='mean',y=fig.index,hue='mean').set(title='Mean of Each Column',xlabel='Mean Value',ylabel='Columns')
# plt.figure(figsize=(10,6),dpi=100)
# sns.histplot(data=df,x='Salary',bins=100,kde=True)
# sns.rugplot(data=df,x='Salary',height=0.5)
```

```python
sns.set_style('ticks')
sns.histplot(data=df,x='Salary',bins=30,color='red',edgecolor='black',linewidth=1.5,alpha=0.9,kde=True)
```

```python
sns.kdeplot(data=df,x='Salary',color='red',alpha=0.5)
```
