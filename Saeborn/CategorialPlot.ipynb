```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df=pd.read_csv('student_exam_data.csv')
sns.boxplot(data=df,x='Test Preparation Course',y='Math Score',hue='Test Preparation Course',palette='Set2').set(title='Math Score by Test Preparation Course',xlabel='Test Preparation Course',ylabel='Math Score')
# df.head()
plt.legend(title='Math Score',bbox_to_anchor=(1.05,1),loc=2,borderaxespad=0.)
```

```python
plt.figure(figsize=(10,6),dpi=200)
sns.violinplot(data=df,x='Reading Score' ,y='Writing Score',inner='stick',gap=0,palette='Set2',split=True ,hue='Reading Score',bw_method=0.5).set(title='Math Score by Test Preparation Course',xlabel='Test Preparation Course',ylabel='Math Score')
sns.move_legend(plt.gca(), "upper left", bbox_to_anchor=(1, 1))
```

```python
sns.swarmplot(data=df,x='Test Preparation Course',size=12,y='Gender',hue='Gender',palette='Set2').set(title='Math Score by Test Preparation Course',xlabel='Test Preparation Course',ylabel='Math Score')
```
