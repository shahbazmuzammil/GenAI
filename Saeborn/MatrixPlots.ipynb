```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv('country_stats.csv')
df=df.set_index('Country')
df
```

```python
sns.heatmap(df.drop('Life expectancy',axis=1),linewidths=0.5,annot=True,cmap='viridis',center=40)
```

```python
sns.clustermap(df.drop('Life expectancy',axis=1),linewidths=0.5,annot=True,cmap='viridis',col_cluster=True)
```

```python
# Load country stats and plot correlation heatmap of numeric columns
df_country = pd.read_csv('country_stats.csv')
display(df_country.head())
# Select only numeric columns before computing correlation
num = df_country.select_dtypes(include=['number'])
sns.heatmap(num.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation matrix - country stats')
plt.show()
```
