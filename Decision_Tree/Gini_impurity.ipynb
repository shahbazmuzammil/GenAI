```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Let's say you want to predict the 'species' of the penguin
X = df[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g']] # 4 features
y = df['species'] # Target variable

# Now check the shapes – they will both have 334 rows!
print(f"Shape of X: {X.shape}")  # Output: (334, 4)
print(f"Shape of y: {y.shape}")  # Output: (334,)

# This will now run without any errors!
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train) 
y_pred = model.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
pd.DataFrame(index=X.columns, data=model.feature_importances_, columns=['Feature Importance']).sort_values(by='Feature Importance', ascending=False)
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12,8),dpi=200)
plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
```

```python
len(X_train)
```

```python
def report_model(model):
    y_pred = model.predict(X_test)
    print(classification_report(y_test, y_pred))
    print(confusion_matrix(y_test, y_pred))
    plt.figure(figsize=(12,8),dpi=200)
    plot_tree(model, feature_names=X.columns, filled=True, rounded=True)
```

```python
report_model(model)
```

```python
prund_tree = DecisionTreeClassifier(max_depth=2)
prund_tree.fit(X_train, y_train)
```

```python
report_model(prund_tree)
```

```python
max_leaf_tree = DecisionTreeClassifier(max_leaf_nodes=3)
```

```python
max_leaf_tree.fit(X_train, y_train)
```

```python
report_model(max_leaf_tree)
```

```python
entropy_tree = DecisionTreeClassifier(criterion='entropy')
entropy_tree.fit(X_train, y_train)
```

```python
report_model(entropy_tree)
```
