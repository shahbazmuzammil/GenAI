```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import pandas as pd
```

```python
image_as_array=mpimg.imread('./premium_photo-1690582417564-b31a9ad96258.avif')
```

```python
plt.imshow(image_as_array)
```

```python
(h,w,c) = image_as_array.shape
print(h,w,c)
```

```python
image_as_array_reshaped = image_as_array.reshape(h*w,c)
print(image_as_array_reshaped.shape)
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=6)
labels=model.fit_predict(image_as_array_reshaped)
```

```python
rgb_codes=model.cluster_centers_
```

```python
rgb_codes[labels]
```

```python
quantize_image= np.reshape(rgb_codes[labels], (h,w,c))
```

```python
plt.imshow(quantize_image)  
```
