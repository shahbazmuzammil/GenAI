```python
    import numpy as np
    import matplotlib.pyplot as plt
    import pandas as pd
    import seaborn as sns
    print("Libraries imported successfully")
```

```python
df = pd.read_csv('Advertising.csv')
```

```python
X = df.drop('sales', axis=1)
y = df['sales']
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
poly_converter = PolynomialFeatures(degree=3, include_bias=False)

poly_features = poly_converter.fit_transform(X)
```

```python
poly_features.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(poly_features, y, test_size=0.3, random_state=101)
   

```

```python
X_train.shape
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
poly_features[0]
```

L2 Regulazrization - Ridge Regression

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, mean_absolute_error
ridge_model= Ridge(alpha=10)
```

```python
ridge_model.fit(X_train,y_train)
ridge_predict = ridge_model.predict(X_test)
MSE = mean_squared_error(y_test,ridge_predict)
RMSE = np.sqrt(MSE)

print("MSE:", MSE)
print("RMSE:", RMSE)
```

Cross Validation

```python
from sklearn.linear_model import RidgeCV
ridge_cv_mode= RidgeCV(alphas=(0.1,1.0,10.0),scoring='neg_mean_squared_error')
ridge_cv_mode.fit(X_train,y_train)



```

```python
ridge_cv_mode.alpha_
```

```python
test_predictions = ridge_cv_mode.predict(X_test)
MEA = mean_absolute_error(y_test, test_predictions)
print("Mean Absolute Error:", MEA)
RMSE = np.sqrt(mean_squared_error(y_test, test_predictions))
print("Root Mean Squared Error:", RMSE)
```

```python
ridge_cv_mode.coef_
```

```python
ridge_cv_mode.best_score_
```

L1 Regularization Lasso Regression


```python
from sklearn.linear_model import LassoCV

lasso_CV_model = LassoCV(eps=0.00001, alphas=10, cv=5, max_iter=10000)
lasso_CV_model.fit(X_train, y_train)
```

```python
lasso_CV_model.alpha_
```

```python
lasso_cv_model_predict = lasso_CV_model.predict(X_test)
```

```python
MAE = mean_absolute_error(y_test,lasso_cv_model_predict)
```

```python
MAE
```

```python
RMSE = np.sqrt(mean_squared_error(y_test, lasso_cv_model_predict))
```
