```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully")
```

```python
blob = pd.read_csv('blobs.csv')
blob.head()
```

```python
sns.scatterplot(data=blob, x='X1', y='X2')
```

```python
moon_cluster = pd.read_csv('cluster_moons.csv')
```

```python
sns.scatterplot(data=moon_cluster, x='X1', y='X2')
```

```python
cluster_circle = pd.read_csv('cluster_circles.csv')
```

```python
sns.scatterplot(data=cluster_circle, x='X1', y='X2')
```

```python
def display_categories(model,data):
    labels = model.fit_predict(data)
    sns.scatterplot(data=data, x='X1', y='X2', hue=labels, palette='Set1')
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3)
```

```python
display_categories(model, blob)
```

```python
model = KMeans(n_clusters=2)
display_categories(model, moon_cluster)
```

```python
model = KMeans(n_clusters=2)
display_categories(model, cluster_circle)
```

```python
from sklearn.cluster import DBSCAN
```

```python
model = DBSCAN()
```

```python
display_categories(model, blob)
```

```python
model = DBSCAN(eps=0.10)
display_categories(model, cluster_circle)
```
