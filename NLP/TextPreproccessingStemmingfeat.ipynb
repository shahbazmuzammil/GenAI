Stemming is a text preprocessing technique in NLP that reduces words to their root of words known as a lemma or base form by chopping off prefixes and suffixes. 

```python
#Classification problem
#Comment of product is a possitive review or negative review
# reviw --> eating ,eat,eaten
```

```python
words = ['eating','eats','eaten','writing','writes','programming','programs','history','final,finalize']
```

```python
#porter Stemmer
from nltk.stem import PorterStemmer
```

PorterStemmer

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
stemming.stem('Congratulations')
```

```python
stemming.stem('sitting')
```

RegexpStemmer Class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowbalStemmer=SnowballStemmer('english')
```

```python
for word in words:
    print (word+"====>",snowbalStemmer.stem(word))
```

```python
stemming.stem('fairly')
```

```python
snowbalStemmer.stem('fairly')
snowbalStemmer.stem('sporting')
```

    Lemmatization

Lemmetizer is to give you the exact form of the word which is meaningful and does not change the meaning 

Wordnet Lemmatizer

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer =  WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
adverb-r
'''
lemmatizer.lemmatize('going',pos='v')
```

```python
words = ['eating','eats','eaten','writing','writes','programming','programs','history','final,finalize']

for word in words:
    print (word+"====>",lemmatizer.lemmatize(word))
```

```python
lemmatizer.lemmatize('goes',pos='v')
```

```python
lemmatizer.lemmatize('fairly',pos='v'),lemmatizer.lemmatize('sportingly')
```
