```python
with open('sample.txt', 'r',encoding='utf-8') as file:
    text_data = file.readline().lower().split()
    uni_words = set(text_data)
```

```python
with open('./prideprejudice.txt', 'r',encoding='utf-8') as file:
    text2_data = file.readline().lower().split()
    uni_words_two = set(text2_data)
```

```python
uni_words_two 
```

```python
all_uni_words = set()

all_uni_words.update(uni_words)
all_uni_words.update(uni_words_two)
```

```python
all_uni_words
```

```python
full_vocab = dict()
i=0

for word in all_uni_words:
    full_vocab[word] = i
    i = i+1
```

```python
full_vocab
```

```python
one_freq =[0]*len(full_vocab)
two_freq =[0]*len(full_vocab)
all_words = ['']*len(full_vocab)
```

```python
two_freq
```

```python
with open('sample.txt', 'r',encoding='utf-8') as file:
    text_data = file.readline().lower().split()
    for word in text_data:
        word_ind = full_vocab[word]
        one_freq[word_ind] += 1
```

```python
one_freq
```

```python
with open('./prideprejudice.txt', 'r',encoding='utf-8') as file:
    text2_data = file.read().lower().split()
    for word in text2_data:
        word_ind = full_vocab[word]
        two_freq[word_ind] += 1
```

```python
two_freq
```

```python
for word in full_vocab:
    ind = full_vocab[word]
    all_words[ind] = word
```
