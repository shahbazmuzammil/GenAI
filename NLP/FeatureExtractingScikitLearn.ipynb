```python
import pandas as pd
import numpy as np
text =['Honored as “rocket force a science and technology experts,',
    '” the Zhang and his a fellow experts ',
    'from private a companies will']
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfTransformer
```

```python
cv = CountVectorizer(stop_words='english')
sparsed_matrix=cv.fit_transform(text)
```

```python
sparsed_matrix.todense()
```

```python
cv.vocabulary_
```

```python
#stop words 
tfidf_transformer = TfidfTransformer()
result= tfidf_transformer.fit_transform(sparsed_matrix)
```

```python
result.todense()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
sparsed_matrix=tfidf_vectorizer.fit_transform(text)
sparsed_matrix.todense()
```
