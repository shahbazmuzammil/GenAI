```python
corpus = """Hello Welcome,to my amazing world's of NLP Deep learning.
Hold for amazing engineering works!. """
print(corpus)
```

```python
##Tokenizatio/n
## sentence into paragraph
from nltk.tokenize import sent_tokenize ,word_tokenize
```

```python
document=sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
#sentence into words
#Tokenization 
#Paragraph ---> words
#sentence --> words
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(corpus))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```
