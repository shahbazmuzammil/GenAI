```python
#Scikit Learn with PCA
```

```python
df = pd.read_csv('./cancer_tumor_data_features.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar = StandardScaler()
```

```python
scaled_X = scalar.fit_transform(df)
```

```python
from sklearn.decomposition import PCA
```

```python
pca_model = PCA(n_components=2)
```

```python
pca_model.fit(scaled_X)
```

```python
pca_model.transform(scaled_X)
```

```python
pc_result=pca_model.fit_transform(scaled_X)
```

```python
plt.scatter(pc_result[:,0],pc_result[:,1],c=cancer_dict['target'])
```

```python
cancer_dict['target']
```

```python
pca_model.components_
```

```python
df_comp = pd.DataFrame(pca_model.components_,index=['PC1','PC2'],columns=df.columns)
```

```python
df_comp 
```

```python
plt.figure(figsize=(20,3),dpi=250)
sns.heatmap(df_comp,annot=True)
```

```python
np.sum(pca_model.explained_variance_ratio_)
```

```python
pca_30 = PCA(n_components=30)
```

```python
pca_30.fit(scaled_X)
```

```python
np.sum(pca_30.explained_variance_ratio_)
```

```python
explained_variance = []

for n in range(1,30):
    pca = PCA(n_components=n)
    pca.fit(scaled_X)
    explained_variance.append(np.sum(pca.explained_variance_ratio_))
```

```python
plt.plot(range(1,30),explained_variance)
plt.xlabel('num of component')
plt.ylabel('variance explained')
```
