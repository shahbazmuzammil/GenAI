````python
import numpy as np
# import matplotlib.pyplot as plt

np.random.seed(42)

# ── Step 1: Create correlated 2D data ──────────────────────────
X = np.random.randn(100, 2)
X[:, 1] = X[:, 0] * 0.9 + X[:, 1] * 0.3   # make correlated
print(f"Data shape: {X.shape}")              # (100, 2)

# ── Step 2: Mean-center the data ───────────────────────────────
X_mean = X - X.mean(axis=0)                 # subtract column means
print(f"Mean after centering: {X_mean.mean(axis=0).round(5)}")  # ~[0, 0]

# ── Step 3: Compute covariance matrix ──────────────────────────
# Cov = (Xᵀ @ X) / (n-1)
n = X_mean.shape[0]
cov = (X_mean.T @ X_mean) / (n - 1)
print(f"Covariance matrix:\n{cov.round(3)}")
# [[1.01  0.93]
#  [0.93  0.95]]

# ── Step 4: Eigendecomposition ─────────────────────────────────
eigenvalues, eigenvectors = np.linalg.eig(cov)
print(f"Eigenvalues:  {eigenvalues.round(3)}")   # [1.91, 0.05]
print(f"Eigenvectors:\n{eigenvectors.round(3)}")

# ── Step 5: Sort by largest eigenvalue ─────────────────────────
idx = np.argsort(eigenvalues)[::-1]             # descending order
eigenvalues  = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
print(f"Sorted eigenvalues: {eigenvalues.round(3)}")

# ── Step 6: Project data onto top-1 principal component ────────
W = eigenvectors[:, :1]                         # top eigenvector (2,1)
X_pca = X_mean @ W                              # (100,2)@(2,1) → (100,1)
print(f"Projected shape: {X_pca.shape}")         # (100, 1)

# ── Step 7: Variance explained ─────────────────────────────────
var_explained = eigenvalues / eigenvalues.sum()
print(f"Variance explained: {(var_explained*100).round(1)}%")
# [97.4%, 2.6%]  → PC1 captures almost everything!
# ```

# ---/

### What each step does
# ```
# Step 1 → raw data         (100, 2)  — 100 samples, 2 features
# Step 2 → center at 0      (100, 2)  — remove mean so origin = data center
# Step 3 → covariance       (2, 2)    — how features vary together
# Step 4 → eigendecomp      values+vectors — find "stretch directions"
# Step 5 → sort             biggest eigenvalue = most variance = PC1
# Step 6 → project          (100, 1)  — compress 2D → 1D along PC1
# Step 7 → explained var    97.4%     — how much info we kept
````
